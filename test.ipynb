```python
import graphviz
import re

# 您提供的目录结构文本
structure_text = """
laplace/
├── Project.toml                     # Julia project dependencies
├── README.md                       # This documentation
├── PROJECT_SUMMARY.md              # Detailed project summary
├── start_demo.jl                   # Interactive demo launcher
├── quick_start.jl                  # Quick start script
├── launch_gui.py                    # GUI launcher
├── laplace_gui.py                   # Main GUI application
├── requirements.txt                # Python dependencies for GUI
│
├── src/                            # Core source code
│   ├── MNISTClassifier.jl             # MNIST classification (2.5k+ lines)
│   ├── GermanCreditClassifier.jl      # German credit classification (1.5k+ lines)
│   ├── BayesianMethods.jl             # Multiple Bayesian methods (900+ lines)
│   ├── OODDatasets.jl                 # OOD dataset loading (800+ lines)
│   ├── OODDetection.jl                # OOD detection algorithms (700+ lines)
│   ├── Visualizations.jl              # Visualization functions (1k+ lines)
│   └── GUIInterface.jl                # GUI backend interface
│
├── test/                           # Comprehensive test suite
│   ├── runtests.jl                    # Main test runner
│   ├── test_mnist.jl                  # MNIST tests (300+ lines)
│   ├── test_german_credit.jl          # German credit tests (200+ lines)  
│   ├── test_bayesian_methods.jl       # Bayesian methods tests (400+ lines)
│   ├── test_ood_datasets.jl           # OOD dataset tests (250+ lines)
│   ├── test_ood_detection.jl          # OOD detection tests (300+ lines)
│   └── test_visualizations.jl         # Visualization tests (350+ lines)
│
├── examples/                       # Usage examples and tutorials
│   ├── mnist_example.jl               # MNIST complete example (200+ lines)
│   ├── german_credit_example.jl       # German credit example (250+ lines)
│   ├── bayesian_methods_comparison.jl # Methods comparison (270+ lines)
│   ├── comprehensive_ood_demo.jl      # Complete OOD demo (400+ lines)
│   ├── mnist_ood_example.jl           # MNIST OOD detection
│   ├── german_credit_ood_example.jl   # German credit OOD detection
│   └── laplace_vs_map_comparison_example.jl # Laplace vs MAP analysis
│
└── Multi-Class-Julia-*.ipynb       # Original reference notebooks
"""

def visualize_directory_structure(text_data: str, output_filename: str = 'laplace_project_structure'):
    """
    解析文本格式的目录结构，并使用 Graphviz 生成可视化图像。

    Args:
        text_data (str): 包含目录结构的字符串。
        output_filename (str): 输出图像的文件名 (不含扩展名)。
    """
    # 初始化一个有向图
    dot = graphviz.Digraph('ProjectStructure', comment='Project Directory Structure')
    dot.attr(rankdir='TB', splines='ortho')  # 设置为从上到下的布局和直角连线
    dot.attr('node', shape='box', style='rounded,filled', fontname='Helvetica')
    dot.attr('edge', color='#444444')

    # parents字典用于追踪每一层的父节点
    parents = {}
    lines = text_data.strip().split('\n')

    for line in lines:
        # 跳过空行或纯连接线
        if not line.strip() or '│' == line.strip().replace(" ", ""):
            continue
        
        # 1. 清理行数据：移除注释和括号内的代码行数
        clean_line = re.sub(r'#.*', '', line).strip()
        clean_line = re.sub(r'\s*\([^)]+\)$', '', clean_line).strip()
        
        # 2. 提取节点名：移除前面的树状结构字符
        node_name_raw = re.sub(r'^[│├└─\s]+', '', clean_line)

        # 3. 计算层级：根据原始行的前缀空格和符号判断
        indentation = len(line) - len(line.lstrip(' │'))
        level = indentation // 4

        # 4. 设置节点样式：文件夹用 folder 图形，文件用 note 图形
        is_directory = node_name_raw.endswith('/')
        node_name = node_name_raw.rstrip('/')

        if is_directory:
            dot.node(node_name, shape='folder', fillcolor='#a9d1f7') # 蓝色文件夹
        else:
            dot.node(node_name, shape='note', fillcolor='#f7f7a9') # 黄色便签

        # 5. 添加节点和边
        if level == 0:
            # 根节点
            parents[0] = node_name
        else:
            # 连接到父节点
            parent_name = parents.get(level - 1)
            if parent_name:
                dot.edge(parent_name, node_name)
        
        # 如果当前节点是目录，更新对应层级的父节点
        if is_directory:
            parents[level] = node_name
            
    # 渲染并保存图像
    try:
        dot.render(output_filename, format='png', view=False, cleanup=True)
        print(f"✅ 可视化成功！图像已保存为 '{output_filename}.png'")
    except graphviz.backend.execute.CalledProcessError:
        print("❌ 错误：无法执行 'dot' 命令。")
        print("请确保您已经安装了 Graphviz (https://graphviz.org/download/) 并且将其添加到了系统的 PATH 环境变量中。")

# --- 运行可视化 ---
visualize_directory_structure(structure_text)
```
